# Qual a nossa vantagem competitiva?

Fui convidado para apresentar um painel no [USP Angels Day](https://www.sympla.com.br/evento/usp-angels-day-ai-first/3478117) em que a ideia é comentar sobre as vantagens competitivas do Brasil no campo de IA **em relação ao mundo**. 

O contexto é _bastante_ desafiador e amplo. Então, quis dar um passo pra trás e tentar responder a seguinte pergunta:

> Hoje, ao avaliarmos o Brasil no cenário de competitividade global de IA?

Ou seja, para podermos falar da vantagem competitiva, primeiro precisamos fazer uma análise para saber em que pé estamos. [^context]

[^context]: A pesquisa e a implementação deste estudo foram feitas com o auxílio do GPT Sol 5.6. A seleção das fontes, a verificação dos resultados e as conclusões são de responsabilidade do autor.

## O que é Vantagem Competitiva?

Para entendermos algo, primeiro precisamos defini-lo. [Em um artigo da HBR](https://hbr.org/1990/03/the-competitive-advantage-of-nations), Michael E. Porter argumenta que a prosperidade nacional é criada, e não herdada. Ela depende da capacidade das indústrias de um país de inovar e aprimorar continuamente sua competitividade.

Um aspecto particularmente interessante do artigo é a definição de quatro pilares que determinam a vantagem competitiva de uma nação:

1. **Condições dos fatores**: a posição do país em relação aos fatores de produção, como mão de obra qualificada e infraestrutura, necessários para competir em determinado setor.

2. **Condições da demanda**: as características da demanda do mercado interno pelos produtos ou serviços daquele setor.

3. **Setores relacionados e de apoio** : a presença ou ausência, no país, de fornecedores e de outros setores relacionados que sejam competitivos internacionalmente.

4. **Estratégia, estrutura e rivalidade das empresas**: as condições existentes no país que influenciam a criação, a organização e a gestão das empresas, assim como a natureza da concorrência no mercado interno.

::: {.callout-note collapse="true"}
É interessante como um artigo escrito nos anos 90 envelheceu tão bem :)
:::

Dito isso, pode-se traçar dois critérios para pautar parte da nossa definição:

- Para sermos competitivos, não dependemos apenas dos recursos disponíveis, mas também da capacidade de criar oportunidades a partir deles.
- Precisamos transformar essas oportunidades em maior capacidade produtiva para o país.

## Medindo Inovação

Medir inovação me parece ser um conceito bastante subjetivo, mas durante as minhas pesquisas, eu encontrei o [Índice Global de Inovação da WIPO](https://www.wipo.int/web-publications/global-innovation-index-2025/en/appendix-i-conceptual-and-measurement-framework-of-the-global-innovation-index.html). Ele separa os insumos que viabilizam a inovação dos resultados produzidos por uma economia, o que está bastante alinhado com o que eu quero apresentar aqui. Consegui transportar a lógica desse índice para o seguinte grafo:

![Dos recursos disponíveis à captura de valor pelo Brasil. Imagem criada com a ajuda do GPT Sol 5.6.](images/competitive-advantage-ai/dag-competitive-advantage-brazil.png){#fig-competitive-advantage-flow}

Assim, a quantidade de pesquisadores seria um **insumo**. Publicações, empresas, produtos, exportações, adoção empresarial e ganhos de produtividade seriam **resultados possíveis**. Para falarmos de vantagem competitiva, porém, não basta que o Brasil produza esses resultados: precisamos medir alguma diferenciação em relação a outros países e, idealmente, capacidade de transformar essa diferença em valor para o país.

A última etapa do grafo (i.e. a captura de valor) é mais difícil de medir (novamente, temos alguma subjetividade aqui). Já os resultados intermediários são um pouco mais fáceis de observar.

# Um baseline

Para termos algum ponto de partida, encontrei o [Vibrancy Tool do HAI, de Stanford](https://hai.stanford.edu/ai-index/global-vibrancy-tool), que consolida métricas e permite comparar países. Com isso, nós conseguimos ter uma noção inicial de onde o Brasil está hoje quando falamos de IA com uma metodologia bastante direta.

## Antes de comparar, precisamos escolher o que comparar

O Vibrancy Tool é um ótimo começo, mas existe uma armadilha aqui: juntar dezenas de métricas em uma nota pode esconder justamente a parte que queremos entender. **Um país pode ter pesquisadores e infraestrutura, mas não convertê-los em empresas**. Pode também ter um mercado enorme, mas capturar pouco do valor gerado.

Por isso, em vez de criar um ranking novo, fixei quatro tipos de evidências e olhei para elas separadamente:

1. **produção na fronteira**, por meio dos modelos notáveis catalogados pelo [Epoch AI](https://epoch.ai/data/ai-models);
2. **presença científica**, usando trabalhos aceitos em sete conferências selecionadas de IA em 2025;
3. **infraestrutura**, com sistemas e capacidade Rmax do [TOP500](https://top500.org/lists/top500/);
4. **conversão econômica**, com investimento privado e empresas de IA financiadas pela primeira vez no [AI Index 2026](https://hai.stanford.edu/ai-index/2026-ai-index-report/economy).

O grupo de comparação também foi definido antes de olharmos os resultados. Selecionamos 16 países: 

- Estados Unidos e China, como referências da fronteira de IA; 
- Canadá, Reino Unido, França e Alemanha, como economias desenvolvidas de referência;
- Argentina, Chile, Colômbia e México, para situar o Brasil na América Latina.

Também incluímos  Índia, Indonésia, África do Sul e Turquia como comparadores estruturais. Assim como o Brasil, são economias do G20 que não estão na fronteira global de IA, mas possuem mercados relevantes, sistemas produtivos e instituições de pesquisa já estabelecidos. A comparação ajuda a verificar se o desempenho brasileiro é apenas uma característica regional ou se ele se diferencia de outras grandes economias emergentes. Por fim, também incluí os Emirados Árabes Unidos entram como referência de uma estratégia recente baseada em investimento intensivo em IA.

Esses países não são idênticos ao Brasil (e nem deveriam ser tratados assim, rs). Como queremos trazer o posicionamento `vs o resto do mundo`, o objetivo aqui é oferecer diferentes pontos de referência.

::: {.callout-note collapse="true"}
Um oitavo indicador compararia a participação dos países em publicações sobre Responsible AI (RAI). Para isso, desenvolvemos um classificador que identificava trabalhos como pertencentes à categoria de RAI a partir de seus títulos e resumos. Ele parecia funcionar em um `dev set`, mas não funcionou no conjunto de teste: obteve 60,0% de precisão e 65,5% de recall. Como o método apresentou `overfitting`, preferi retirar o indicador deste primeiro estudo do que produzir uma comparação entre países que não poderíamos confiar.
:::

## O primeiro retrato

Vamos começar pela produção científica. Reconstruímos o universo de trabalhos aceitos em AAAI, ACL, ICML, NeurIPS, EMNLP, ICLR e no track Applied Data Science da KDD em 2025. Para evitar que uma conferência muito grande dominasse o resultado, o indicador principal é a média da participação ponderada do país, dando o mesmo peso a cada venue.

![Participação fracionária média por país nas sete conferências. Cada venue recebe o mesmo peso.](images/competitive-advantage-ai/baseline-conference-presence-2025.png){#fig-conference-baseline}

O Brasil aparece em **10º de 16 países**, com **0,0640%** da participação fracionária média. É uma presença pequena, embora seja a segunda maior entre os países latino-americanos deste painel.

É importante sermos precisos: isso **não** mede toda a ciência brasileira em IA. Não inclui todos os periódicos, conferências regionais, impacto por citações, liderança de autoria ou especializações. Contudo, o resultado contesta a ideia de uma presença ampla **nestas conferências**, mas não encerra a hipótese sobre qualidade ou eficiência científica.

Quando olhamos para modelos e infraestrutura, a história fica mais interessante — e também mais difícil de resumir em dois números:

![Modelos notáveis atribuídos a organizações do país em 2025 e capacidade Rmax no TOP500 de junho de 2026.](images/competitive-advantage-ai/baseline-frontier-infrastructure.png){#fig-frontier-infrastructure}

No snapshot do Epoch, há 103 modelos notáveis em 2025 e nenhum deles é atribuído a uma organização brasileira. O mesmo continua verdadeiro no recorte de 2026 até a data da coleta. Isso é evidência contrária à presença brasileira **dentro da fronteira observada pelo Epoch** — não uma prova de que o país não produza modelos relevantes.

O [Sabiá 4, da Maritaca](https://www.maritaca.ai/post/sabia-4), ajuda a entender essa diferença. Ele não aparece no catálogo do Epoch, mas foi desenvolvido com foco em português e em contextos brasileiros. Na [avaliação publicada pela própria empresa](https://docs.maritaca.ai/pt/introducao), o Sabiá 4 Thinking apresenta resultados próximos ou superiores ao Opus 4.8 em algumas tarefas brasileiras, além de um custo menor para executar a suíte avaliada. Parte desses benchmarks, porém, foi construída internamente. O caso é uma evidência promissora de especialização local, mas ainda não permite concluir que o Brasil esteja competindo de forma ampla na fronteira global.

Também temos o [Tucano 2](https://arxiv.org/abs/2603.03543), um modelo voltado ao português brasileiro e, entre os casos discutidos aqui, o único totalmente aberto. O projeto foi liderado pelo pesquisador brasileiro Nicholas Kluge Corrêa na [Universidade de Bonn, na Alemanha](https://www.uni-bonn.de/en/university/press-and-communications/publications/forsch/forsch-2025-02-english/forsch-content-2025-02/ki-chatbots-fuer-alle-sprachen). Podemos considerá-lo um modelo brasileiro pelo idioma e pelo problema que procura resolver, mas seu desenvolvimento também revela nossa dependência: o treinamento foi viabilizado por uma universidade e por infraestrutura localizadas fora do Brasil.

Já no TOP500, o Brasil tinha **10 sistemas e 124,52 PFlop/s** em novembro de 2025; em junho de 2026, continuava com 10 sistemas e passou a **143,12 PFlop/s**. É uma posição intermediária no painel, bem diferente do zero observado em modelos notáveis.

A existência desses sistemas importa. Ela mostra que algumas organizações brasileiras conseguiram investir e operar infraestrutura pertencente a um grupo relativamente seleto. Mas capacidade instalada e capacidade acessível não são a mesma coisa. O TOP500 não informa quem pode utilizar esses computadores, por quanto tempo, a que custo ou para quais tipos de projeto.

Essa diferença fica ainda mais clara quando olhamos para a nuvem. A [Magalu Cloud](https://docs.magalu.cloud/docs/computing/virtual-machine/instance-types-and-pricing/instance-types) oferece instâncias com GPUs L40 e L40S em infraestrutura nacional. São equipamentos bastante capazes para executar modelos, ajustar modelos menores e desenvolver aplicações. Já a [H100 oferecida pela AWS](https://aws.amazon.com/about-aws/whats-new/2025/08/amazon-p5-single-gpu-instances-now-available/) inclusive em São Paulo foi projetada para o treinamento pesado de IA.

Para termos uma noção da diferença sem precisarmos virar especialistas em hardware: uma H100 possui cerca de **1,7 vez a memória**, movimenta esses dados quase **quatro vezes mais rápido** e pode entregar mais de **cinco vezes** o processamento especializado de uma L40 em determinadas operações usadas por modelos de IA. Mais importante do que comparar uma placa com outra, porém, é observar quantas delas podem trabalhar juntas. Uma empresa pode alugar uma ou algumas GPUs em uma nuvem nacional. Laboratórios na fronteira projetam supercomputadores inteiros ao redor delas.

A escala da OpenAI ajuda a visualizar esse abismo. Em 2025, a empresa informou ter aproximadamente **1,9 gigawatt de compute disponível**. O Stargate, sua plataforma de infraestrutura, já descreve supercomputadores com **mais de 100 mil GPUs** conectadas e projetos de expansão medidos em milhões de chips. Esses números misturam capacidade operacional e projetos em desenvolvimento, portanto não representam uma contagem simples de GPUs que a empresa possa usar hoje. Ainda assim, mostram que a competição na fronteira ocorre na escala de data centers, energia e redes — não apenas na compra de uma placa mais rápida.

Usar um provedor global pode ampliar o acesso à melhor infraestrutura disponível. Ao mesmo tempo, transfere parte do controle para uma empresa estrangeira. É aqui que a discussão deixa de ser apenas sobre capacidade e passa a incluir **soberania computacional**: onde os dados são processados, qual legislação se aplica, quem opera a infraestrutura e quantos fornecedores podem ser substituídos.

A Europa oferece um contraponto interessante. Ela não precisa produzir internamente todos os componentes da cadeia para buscar maior autonomia. Um [post de Jarkko Syrjälä no LinkedIn](https://www.linkedin.com/posts/syrjala_it-sounds-like-a-geopolitical-joke-european-share-7487853204421021696-0kYl/) resume essa aparente contradição usando um modelo chinês executado por uma empresa europeia em infraestrutura localizada na Finlândia:

![Publicação de Jarkko Syrjälä sobre soberania europeia em IA. Captura fornecida pelo autor deste post.](images/competitive-advantage-ai/linkedin-ai-sovereignty-finland.jpg){#fig-linkedin-ai-sovereignty width=70%}

O exemplo é informal, mas a distinção é útil: soberania não significa autarquia tecnológica. Significa preservar opções sobre onde os modelos são executados, onde os dados permanecem e de quais fornecedores dependemos. A própria [Nebius documenta processamento de dados em seu data center na Finlândia](https://docs.tokenfactory.nebius.com/data-lab/data-processing), enquanto a [União Europeia investe em uma rede de *AI Factories*](https://digital-strategy.ec.europa.eu/en/policies/ai-factories) aberta a pesquisadores, startups e empresas.

Então, a infraestrutura brasileira continua sendo uma evidência **mista**. Temos sinais de investimento concentrado, modelos locais especializados e alguma oferta nacional de nuvem. Ainda não demonstramos, porém, que essa capacidade seja suficientemente avançada, distribuída e acessível para sustentar inovação em escala.

## E a conversão econômica?

O [AI Index 2026](https://hai.stanford.edu/ai-index/2026-ai-index-report) publica os 15 primeiros países em investimento privado em IA e em empresas financiadas pela primeira vez em 2025. O Brasil não aparece em nenhuma das duas listas. Como a fonte não publica o valor brasileiro, o resultado correto não é zero: ele é **censurado**.

![A linha azul representa o intervalo possível para o Brasil; a seta marca o limite superior para que o país permanecesse fora do top 15 publicado.](images/competitive-advantage-ai/baseline-economic-conversion-2025.png){#fig-economic-conversion}

Isso nos permite afirmar apenas que o investimento foi inferior a **US$ 0,970 bilhão** e que foram financiadas pela primeira vez **menos de 33 empresas**. É suficiente para contestar uma posição econômica globalmente destacada em 2025, mas não para estimar quanto foi investido no Brasil — e muito menos para atribuir esse valor a setores específicos.

O baseline, então, fica assim:

| Família | Resultado brasileiro | Leitura |
|---|---:|---|
| Modelos notáveis | 0 em 2025 e 0 em 2026 YTD | contrária à presença observada na fronteira |
| Conferências selecionadas | 0,0640%; 10º/16 | contrária à liderança científica ampla |
| TOP500 | 10 sistemas; 143,12 PFlop/s em jun. 2026 | mista |
| Investimento privado | < US$ 0,970 bi em 2025 | contrária à liderança econômica ampla |
| Novas empresas financiadas | < 33 em 2025 | contrária à conversão ampla |

A hipótese de que o desempenho brasileiro seria superior ao esperado entre emergentes semelhantes fica **contestada** neste baseline. A conclusão não depende de uma nota agregada: produção na fronteira, ciência selecionada e conversão econômica não mostram superioridade robusta; infraestrutura é a exceção parcial, mas ainda sem evidência de acesso e uso.

# Então onde pode estar a vantagem?

A ausência de uma liderança ampla não responde à pergunta original por completo. Porter não diz que um país precisa vencer em tudo; a vantagem pode aparecer na combinação entre fatores, demanda e setores específicos. O próximo passo é procurar mecanismos que sejam plausíveis **e** mensuráveis, sem transformar bons exemplos em conclusão nacional.

## P&D: pouco para a fronteira, mas não pouco entre os pares

Segundo o [World Development Indicators](https://data.worldbank.org/indicator/GB.XPD.RSDV.GD.ZS?locations=BR), o Brasil investiu **1,19% do PIB em P&D em 2023**. No nosso painel, isso o coloca em 9º de 16: abaixo de Estados Unidos, China e das referências desenvolvidas, mas acima dos demais países latino-americanos observados e de Índia, África do Sul e Indonésia. A Turquia, com 1,42%, está à frente.

Ou seja, chamar o investimento brasileiro simplesmente de baixo seria esconder metade da história. Ele é limitado diante da fronteira, mas relativamente forte entre vários pares. Ao mesmo tempo, esse insumo intermediário convive com resultados modestos nas conferências, zero modelos notáveis observados e conversão econômica fora do top 15.

Isso torna a hipótese de P&D **mista** e fortalece outra pergunta: nosso problema é somente colocar mais recursos, ou também conectá-los melhor? Ainda não conseguimos separar P&D público e empresarial comparável nem identificar a parcela específica de IA. Os anos disponíveis também variam entre países. Portanto, não temos uma estimativa causal de eficiência.

## Ciência e infraestrutura: duas lacunas diferentes

A evidência científica continua **inconclusiva** para a hipótese completa. O resultado nas sete conferências é fraco, mas não mede impacto, especialização ou produção por pesquisador. Podemos dizer que não encontramos liderança ampla naquele recorte; não podemos dizer que testamos toda a ciência brasileira em IA.

Na infraestrutura, a conclusão permanece **mista**. Existe capacidade HPC nominal que coloca o Brasil no meio do painel, mas faltam os elos decisivos: aceleradores adequados, acesso efetivo, custo e uso. É uma diferença importante. Possuir um sistema e conseguir transformá-lo em experimentos, modelos e produtos são etapas distintas do nosso grafo.

## Demanda e setores: três hipóteses que valem continuar testando

O Brasil tinha cerca de **212,8 milhões de habitantes em 2025**, segundo o Banco Mundial. Escala, por si só, não é vantagem; Índia, China e Indonésia tornam isso bastante evidente. Ela passa a importar quando vem acompanhada de dados próprios, problemas locais, rivalidade e empresas capazes de aprender com o mercado.

Encontrei três candidatos que ilustram esse mecanismo:

- **Finanças.** O Nubank atribui parte da evolução do crédito ao seu modelo proprietário `nuFormer` e reportou mais de 10 milhões de usuários mensais do Pix com IA em seus [resultados de 2025](https://international.nubank.com.br/pt-br/companhia/nu-holdings-ltd-divulga-resultados-financeiros-do-quarto-trimestre-e-do-ano-de-2025/). O Itaú documenta uma plataforma generativa combinando dados e modelos proprietários, além de ter iniciado seu [agente de investimentos](https://www.itau.com.br/relacoes-com-investidores/noticias/itau-lanca-agente-de-investimentos-baseado-em-inteligencia-artificial-generativa/) com 10 mil clientes.
- **Jurídico.** O [Jus IA](https://ia.jusbrasil.com.br/) combina um produto especializado com o acervo jurídico local e divulga métricas de uso e satisfação. Um [estudo experimental associado ao produto](https://arxiv.org/abs/2510.18108) reporta vantagem sobre sistemas generalistas, mas não é uma validação independente.
- **Português e especialização de domínio.** A Maritaca desenvolve modelos voltados ao português, e sua relação com o Jusbrasil mostra um caminho concreto de pesquisa e modelo local para uma aplicação vertical.

Esses casos ajudam? **Sim, como hipóteses.** Eles mostram mecanismos reais: dados transacionais, acervo jurídico, idioma e distribuição local. Mas ainda não temos comparadores internacionais equivalentes, contrafactuais, receitas específicas, custo total, dependência de cloud e modelos externos, exportações ou captura local verificável. Declarações corporativas podem provar que um produto existe; não provam, sozinhas, que ele constitui uma vantagem competitiva do país.

Por isso, classifico finanças, jurídico e modelos em português como **vantagens candidatas**, e não como vantagens demonstradas.

# O principal sinal está nas conexões

Voltando ao nosso grafo, há um padrão descritivo difícil de ignorar: P&D e HPC aparecem em posições intermediárias, enquanto modelos notáveis, conferências selecionadas e indicadores econômicos ficam bem atrás. Ao mesmo tempo, encontramos empresas locais tentando conectar dados, domínio e distribuição a produtos de IA.

A hipótese de que o gargalo brasileiro está na **conversão** de insumos em pesquisa, produtos, escala e valor local recebe suporte provisório. Provisório é a palavra importante: as fontes têm universos diferentes e não identificam causalidade. Ainda assim, essa explicação organiza melhor a evidência do que a ideia de que simplesmente não temos recursos ou oportunidades.

E o que podemos responder à pergunta inicial?

> Com a evidência disponível, não demonstramos uma vantagem competitiva ampla do Brasil em IA. Encontramos capacidade intermediária e mecanismos setoriais promissores, especialmente em finanças, jurídico e português, mas ainda não a combinação completa de diferenciação, resultado e captura local necessária para chamá-los de vantagens competitivas nacionais.

É uma conclusão menos empolgante do que começar com uma lista de setores em que "o Brasil é forte". Na minha opinião, ela também é mais útil. Se a oportunidade está nas conexões, o debate deixa de ser apenas quantos pesquisadores, GPUs ou bilhões temos e passa a incluir quais transições estão falhando — e quais casos conseguem atravessá-las.

## Limitações e próximos testes

Este é um estudo exploratório, não um censo de todo o ecossistema. As conferências representam um recorte; o Epoch possui uma taxonomia própria; o TOP500 mede HPC nominal; e os dados econômicos publicados censuram países fora do top 15. As observações de P&D têm anos diferentes, e os estudos de caso dependem parcialmente de informações das próprias empresas.

Também deixamos três perguntas abertas, agora de forma explícita:

1. A presença científica muda quando medimos impacto, liderança e especializações brasileiras, normalizando por pesquisadores e investimento?
2. Quanto da infraestrutura nominal é realmente acessível para IA, e quem consegue utilizá-la?
3. Nubank, Itaú, Jusbrasil e Maritaca geram produtividade, exportação e captura local acima de comparadores — ou são bons casos de adoção em um mercado grande?

Os números, contratos de decisão, checksums, scripts e testes desta análise estão no diretório [`experiments/ai-opportunities-in-brazil`](https://github.com/abarbosa94/personal_blog/tree/main/experiments/ai-opportunities-in-brazil). A coleta foi congelada em 29–30 de julho de 2026; 2026 é apresentado como YTD quando aplicável. Assim conseguimos voltar a essas perguntas sem mover a régua depois de conhecer o resultado.
